```python
# Importing Libraries
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms
```

```python
# Importing MNIST Dataset from torchvision
dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transforms.ToTensor())
# Passing the Dataset to DataLoader
x = torch.utils.data.DataLoader(dataset, batch_size=64, shuffle=True)
# Taking single batch
x_single = next(iter(x))
x_images, x_labels = x_single
print(x_images.shape)
```

```python
# Defining my very first neural network
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(28*28,128)
        self.fc2 = nn.Linear(128,10)

    def forward(self, x):
        x = torch.flatten(x,1)
        x = self.fc1(x)
        x = F.relu(x)
        x = self.fc2(x)
        return x

my_nn = Net()
print(my_nn)
```

```python
result = my_nn(x_images)
print(result)
```
